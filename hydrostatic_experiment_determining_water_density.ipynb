# Determining water density from a mass–volume experiment

This tutorial generates a deterministic synthetic dataset, fits the physical model $M(V)=M_0+\rho(V-V_0)$ by ordinary least squares, and reports the slope as the water-density estimate. OLS treats volume as the controlled predictor; the displayed coordinate uncertainties are not regression weights.

```python
from class_water_density import WaterDensity

# Define the reference state used to generate the teaching dataset.
wd = WaterDensity(M0=200, V0=100, rho=0.997)

# Step 1: Generate fake experimental data
df = wd.gen_fake_data(
    n_points=10,  # number of data points
    V_min=110,    # minimum volume
    V_max=200,    # maximum volume
    err_M=0.02,  # relative error in mass
    err_V=0.02,   # relative error in volume
    noise=2,      # half-width of the uniform mass perturbation
    random_state=42,  # makes the example exactly reproducible
)

# Step 2: Perform least squares linear regression on the generated data
tb1, fit_results = wd.calculate_fit()

# Step 3: Format the data table for LaTeX-like reporting
tb2 = wd.format_table()

tb1
```
